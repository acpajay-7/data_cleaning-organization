```python
import pandas as pd

file_path = '/content/drive/MyDrive/data/indian_businessman_2026.xlsx'

df = pd.read_excel(file_path)

df.head()
```

```python
df.info()
```

```python
df_clean = df.copy()
```

```python
empty_rows = df.isna().all(axis=1)
empty_rows.sum()
```

```python
empty_caluman = df.isna().all(axis=0)
empty_caluman.sum()
```

```python
df = df.dropna(how='all', axis=1)
```

```python
df.info()
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
    .str.replace(r'[^\w]', '', regex=True)
)
```

```python
df.columns
```

```python
df = df.drop(columns=['mobile_phone'])
```

```python
df.info()
```

```python
df['corporate_phone'] = df['corporate_phone'].fillna('unknown')
```

```python
df['state'] = df['state'].fillna('unknown')
```

```python
df['company_city'] = df['company_city'].fillna('unknown')
```

```python
df['company_state'] = df['company_state'].fillna('unknown')
```

```python
df['company_country'] = df['company_country'].fillna('unknown')
```

```python
df['company_phone'] = df['company_phone'].fillna('0000000000')
```

```python
df['secondary_email_status'] = df['secondary_email_status'].fillna('na')
```

```python
df['secondary_email_verification_source'] = df['secondary_email_verification_source'].fillna('na')
```

```python
df.info()
```

```python
p = ['first_name', 'last_name', 'title', 'company_name_for_emails', 'industry', 'state', 'country', 'company_address', 'company_city', 'company_state', 'company_country', 'company_phone', 'secondary_email_status', 'secondary_email_verification_source']
for col in p :
  df[col] = df[col].astype(str).str.strip().str.lower()
```

```python
df.head(10)
```

```python
df.to_excel('/content/drive/MyDrive/data/cleaned_data4.xlsx',index=False)
```
